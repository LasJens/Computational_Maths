## Задача VI.8.3

Даны узлы интерполяции в градусах: $0^\circ$, $30^\circ$, $45^\circ$, $60^\circ$.  
Переведём в радианы:  
$x_0 = 0$, $x_1 = \frac{\pi}{6}$, $x_2 = \frac{\pi}{4}$, $x_3 = \frac{\pi}{3}$.  
Точка интерполяции: $x = 5^\circ = \frac{\pi}{36}$.  
Функция $f(x) = \sin x$.

Погрешность интерполяционного многочлена Лагранжа степени $n$:  
$$
R_n(x) = \frac{f^{(n+1)}(\xi)}{(n+1)!} \cdot \omega_{n+1}(x),
$$
где $\omega_{n+1}(x) = (x - x_0)(x - x_1)\dots(x - x_n)$, $\xi \in [\min(x_i), \max(x_i)]$.

---

**а) Линейная интерполяция ($n=1$)**  
Узлы: $x_0 = 0$, $x_1 = \frac{\pi}{6}$.

$$
\omega_2(x) = (x - 0)\left(x - \frac{\pi}{6}\right),
$$
$$
\omega_2\left(\frac{\pi}{36}\right) = \frac{\pi}{36} \cdot \left( \frac{\pi}{36} - \frac{\pi}{6} \right) = \frac{\pi}{36} \cdot \left( -\frac{5\pi}{36} \right) = -\frac{5\pi^2}{1296}.
$$
Модуль: $|\omega_2| = \frac{5\pi^2}{1296}$.

$f''(x) = -\sin x$, $|\sin \xi| \le 1 \Rightarrow |f''(\xi)| \le 1$.

$$
|R_1| \le \frac{1}{2!} \cdot \frac{5\pi^2}{1296} = \frac{5\pi^2}{2592}.
$$
При $\pi \approx 3.14159265$:  
$\pi^2 \approx 9.8696044$,  
$5\pi^2 \approx 49.348022$,  
$\frac{49.348022}{2592} \approx 0.01904$.

---

**б) Квадратичная интерполяция ($n=2$)**  
Узлы: $x_0 = 0$, $x_1 = \frac{\pi}{6}$, $x_2 = \frac{\pi}{4}$.

$$
\omega_3(x) = (x - 0)\left(x - \frac{\pi}{6}\right)\left(x - \frac{\pi}{4}\right),
$$
$$
\omega_3\left(\frac{\pi}{36}\right) = \frac{\pi}{36} \cdot \left( -\frac{5\pi}{36} \right) \cdot \left( \frac{\pi}{36} - \frac{\pi}{4} \right).
$$
$$
\frac{\pi}{36} - \frac{\pi}{4} = \frac{\pi - 9\pi}{36} = -\frac{8\pi}{36} = -\frac{2\pi}{9}.
$$
$$
\omega_3\left(\frac{\pi}{36}\right) = \frac{\pi}{36} \cdot \frac{5\pi}{36} \cdot \frac{2\pi}{9} = \frac{10\pi^3}{11664}.
$$
Модуль: $|\omega_3| = \frac{10\pi^3}{11664}$.

$f'''(x) = -\cos x$, $|\cos \xi| \le 1 \Rightarrow |f'''(\xi)| \le 1$.

$$
|R_2| \le \frac{1}{3!} \cdot \frac{10\pi^3}{11664} = \frac{10\pi^3}{69984}.
$$
$\pi^3 \approx 31.00627668$,  
$10\pi^3 \approx 310.0627668$,  
$\frac{310.0627668}{69984} \approx 0.004431$.

---

**в) Кубическая интерполяция ($n=3$)**  
Узлы: $x_0 = 0$, $x_1 = \frac{\pi}{6}$, $x_2 = \frac{\pi}{4}$, $x_3 = \frac{\pi}{3}$.

$$
\omega_4(x) = (x - 0)\left(x - \frac{\pi}{6}\right)\left(x - \frac{\pi}{4}\right)\left(x - \frac{\pi}{3}\right),
$$
$$
\omega_4\left(\frac{\pi}{36}\right) = \frac{\pi}{36} \cdot \left( -\frac{5\pi}{36} \right) \cdot \left( -\frac{2\pi}{9} \right) \cdot \left( \frac{\pi}{36} - \frac{\pi}{3} \right).
$$
$$
\frac{\pi}{36} - \frac{\pi}{3} = \frac{\pi - 12\pi}{36} = -\frac{11\pi}{36}.
$$
$$
\omega_4\left(\frac{\pi}{36}\right) = \frac{\pi}{36} \cdot \frac{5\pi}{36} \cdot \frac{2\pi}{9} \cdot \frac{11\pi}{36} = \frac{110\pi^4}{1259712}.
$$
Модуль: $|\omega_4| = \frac{110\pi^4}{1259712}$.

$f^{(4)}(x) = \sin x$, $|\sin \xi| \le 1 \Rightarrow |f^{(4)}(\xi)| \le 1$.

$$
|R_3| \le \frac{1}{4!} \cdot \frac{110\pi^4}{1259712} = \frac{110\pi^4}{30233088}.
$$
$\pi^4 \approx 97.40909103$,  
$110\pi^4 \approx 10714.99999$,  
$\frac{10715.0}{30233088} \approx 0.0003544$.

---

Ответ:  
Линейная интерполяция: $\approx 1.9 \cdot 10^{-2}$,  
Квадратичная интерполяция: $\approx 4.4 \cdot 10^{-3}$,  
Кубическая интерполяция: $\approx 3.5 \cdot 10^{-4}$.

## Задача VI.8.14

Даны узлы интерполяции: $x_0 = 0$, $x_1 = \frac{\pi}{6}$, $x_2 = \frac{\pi}{4}$, $x_3 = \frac{\pi}{3}$.  
Точка интерполяции: $x = \frac{7\pi}{24}$.  
Значения функции в узлах заданы с погрешностью $\delta = 10^{-2}$.

Погрешность интерполяции складывается из:
1) Погрешности метода $R_n(x)$
2) Погрешности исходных данных $\Delta f(x)$

---

**1. Погрешность метода (кубическая интерполяция)**

Для многочлена Лагранжа степени 3:
$$R_3(x) = \frac{f^{(4)}(\xi)}{4!} \cdot \omega_4(x),$$
где $\omega_4(x) = (x-x_0)(x-x_1)(x-x_2)(x-x_3)$.

Вычислим $\omega_4\left(\frac{7\pi}{24}\right)$:
$$x - x_0 = \frac{7\pi}{24} - 0 = \frac{7\pi}{24},$$
$$x - x_1 = \frac{7\pi}{24} - \frac{\pi}{6} = \frac{7\pi}{24} - \frac{4\pi}{24} = \frac{3\pi}{24} = \frac{\pi}{8},$$
$$x - x_2 = \frac{7\pi}{24} - \frac{\pi}{4} = \frac{7\pi}{24} - \frac{6\pi}{24} = \frac{\pi}{24},$$
$$x - x_3 = \frac{7\pi}{24} - \frac{\pi}{3} = \frac{7\pi}{24} - \frac{8\pi}{24} = -\frac{\pi}{24}.$$

$$\omega_4\left(\frac{7\pi}{24}\right) = \frac{7\pi}{24} \cdot \frac{\pi}{8} \cdot \frac{\pi}{24} \cdot \left(-\frac{\pi}{24}\right) = -\frac{7\pi^4}{24 \cdot 8 \cdot 24 \cdot 24} = -\frac{7\pi^4}{110592}.$$

Модуль: $|\omega_4| = \frac{7\pi^4}{110592}$.

Для $f(x) = \sin x$: $f^{(4)}(x) = \sin x$, $|\sin \xi| \le 1$.

$$|R_3| \le \frac{1}{24} \cdot \frac{7\pi^4}{110592} = \frac{7\pi^4}{2654208}.$$

При $\pi \approx 3.14159265$:
$\pi^4 \approx 97.40909103$,
$7\pi^4 \approx 681.8636372$,
$\frac{681.8636372}{2654208} \approx 0.000257$.

---

**2. Погрешность исходных данных**

Погрешность интерполяции из-за ошибок в значениях функции:
$$\Delta I(x) = \sum_{j=0}^3 |\ell_j(x)| \cdot \delta,$$
где $\ell_j(x)$ - базисные полиномы Лагранжа.

Вычислим $\ell_j\left(\frac{7\pi}{24}\right)$:

Общая формула:
$$\ell_j(x) = \prod_{\substack{k=0 \\ k \ne j}}^3 \frac{x - x_k}{x_j - x_k}.$$

Для $j=0$:
$$\ell_0 = \frac{(x-x_1)(x-x_2)(x-x_3)}{(x_0-x_1)(x_0-x_2)(x_0-x_3)} = \frac{\frac{\pi}{8} \cdot \frac{\pi}{24} \cdot \left(-\frac{\pi}{24}\right)}{\left(-\frac{\pi}{6}\right)\left(-\frac{\pi}{4}\right)\left(-\frac{\pi}{3}\right)} = \frac{-\frac{\pi^3}{4608}}{-\frac{\pi^3}{72}} = \frac{1}{64} = 0.015625.$$

Для $j=1$:
$$\ell_1 = \frac{(x-x_0)(x-x_2)(x-x_3)}{(x_1-x_0)(x_1-x_2)(x_1-x_3)} = \frac{\frac{7\pi}{24} \cdot \frac{\pi}{24} \cdot \left(-\frac{\pi}{24}\right)}{\frac{\pi}{6} \cdot \left(-\frac{\pi}{12}\right) \cdot \left(-\frac{\pi}{6}\right)} = \frac{-\frac{7\pi^3}{13824}}{-\frac{\pi^3}{432}} = \frac{7}{32} = 0.21875.$$

Для $j=2$:
$$\ell_2 = \frac{(x-x_0)(x-x_1)(x-x_3)}{(x_2-x_0)(x_2-x_1)(x_2-x_3)} = \frac{\frac{7\pi}{24} \cdot \frac{\pi}{8} \cdot \left(-\frac{\pi}{24}\right)}{\frac{\pi}{4} \cdot \frac{\pi}{12} \cdot \left(-\frac{\pi}{12}\right)} = \frac{-\frac{7\pi^3}{4608}}{-\frac{\pi^3}{576}} = \frac{7}{8} = 0.875.$$

Для $j=3$:
$$\ell_3 = \frac{(x-x_0)(x-x_1)(x-x_2)}{(x_3-x_0)(x_3-x_1)(x_3-x_2)} = \frac{\frac{7\pi}{24} \cdot \frac{\pi}{8} \cdot \frac{\pi}{24}}{\frac{\pi}{3} \cdot \frac{\pi}{6} \cdot \frac{\pi}{12}} = \frac{\frac{7\pi^3}{4608}}{\frac{\pi^3}{216}} = \frac{21}{64} = 0.328125.$$

Сумма модулей:
$$\sum_{j=0}^3 |\ell_j| = 0.015625 + 0.21875 + 0.875 + 0.328125 = 1.4375.$$

Погрешность от исходных данных:
$$\Delta I = 1.4375 \cdot 10^{-2} \approx 0.014375.$$

---

**3. Суммарная погрешность**

Полная погрешность:
$$\Delta_{\text{полн}} = |R_3| + \Delta I \approx 0.000257 + 0.014375 = 0.014632.$$

---

Ответ: $\approx 1.46 \cdot 10^{-2}$

## Задача VI.9.1

Дана таблица значений функции $f(x) = 4x - \cos x$:

| $x$    | $f(x)$  |
|--------|---------|
| 0.0    | -1.000  |
| 0.1    | -0.595  |
| 0.3    | 0.245   |
| 0.5    | 1.122   |

Требуется найти корень уравнения $4x - \cos x = 0$ методом обратной интерполяции.

---

### 1. Построение обратной интерполяции

Будем искать $x$ как функцию от $y$: $x = g(y)$, где $y = f(x)$.  
В узловых точках:

| $y$    | $x$    |
|--------|--------|
| -1.000 | 0.0    |
| -0.595 | 0.1    |
| 0.245  | 0.3    |
| 1.122  | 0.5    |

Ищем корень при $y = 0$.

---

### 2. Кубическая интерполяция для обратной функции

Используем интерполяционный многочлен Лагранжа степени 3:

$$x = g(y) = \sum_{j=0}^3 x_j \cdot \ell_j(y),$$

где базисные полиномы:

$$\ell_j(y) = \prod_{\substack{k=0 \\ k \ne j}}^3 \frac{y - y_k}{y_j - y_k}.$$

---

**Вычисление базисных полиномов в точке $y=0$:**

Для $j=0$ ($y_0 = -1.000$, $x_0 = 0.0$):

$$\ell_0(0) = \frac{(0 - y_1)(0 - y_2)(0 - y_3)}{(y_0 - y_1)(y_0 - y_2)(y_0 - y_3)} = \frac{(0 - (-0.595))(0 - 0.245)(0 - 1.122)}{(-1.000 - (-0.595))(-1.000 - 0.245)(-1.000 - 1.122)}$$

$$= \frac{(0.595)(-0.245)(-1.122)}{(-0.405)(-1.245)(-2.122)} = \frac{0.1635}{(-1.070)} \approx -0.1528.$$

---

Для $j=1$ ($y_1 = -0.595$, $x_1 = 0.1$):

$$\ell_1(0) = \frac{(0 - y_0)(0 - y_2)(0 - y_3)}{(y_1 - y_0)(y_1 - y_2)(y_1 - y_3)} = \frac{(0 - (-1.000))(0 - 0.245)(0 - 1.122)}{(-0.595 - (-1.000))(-0.595 - 0.245)(-0.595 - 1.122)}$$

$$= \frac{(1.000)(-0.245)(-1.122)}{(0.405)(-0.840)(-1.717)} = \frac{0.2749}{0.584} \approx 0.4707.$$

---

Для $j=2$ ($y_2 = 0.245$, $x_2 = 0.3$):

$$\ell_2(0) = \frac{(0 - y_0)(0 - y_1)(0 - y_3)}{(y_2 - y_0)(y_2 - y_1)(y_2 - y_3)} = \frac{(0 - (-1.000))(0 - (-0.595))(0 - 1.122)}{(0.245 - (-1.000))(0.245 - (-0.595))(0.245 - 1.122)}$$

$$= \frac{(1.000)(0.595)(-1.122)}{(1.245)(0.840)(-0.877)} = \frac{-0.6676}{-0.916} \approx 0.7288.$$

---

Для $j=3$ ($y_3 = 1.122$, $x_3 = 0.5$):

$$\ell_3(0) = \frac{(0 - y_0)(0 - y_1)(0 - y_2)}{(y_3 - y_0)(y_3 - y_1)(y_3 - y_2)} = \frac{(0 - (-1.000))(0 - (-0.595))(0 - 0.245)}{(1.122 - (-1.000))(1.122 - (-0.595))(1.122 - 0.245)}$$

$$= \frac{(1.000)(0.595)(-0.245)}{(2.122)(1.717)(0.877)} = \frac{-0.1458}{3.194} \approx -0.0456.$$

---

### 3. Вычисление корня

$$x \approx \sum_{j=0}^3 x_j \cdot \ell_j(0) = 0.0 \cdot (-0.1528) + 0.1 \cdot 0.4707 + 0.3 \cdot 0.7288 + 0.5 \cdot (-0.0456)$$

$$= 0 + 0.04707 + 0.21864 - 0.0228 = 0.24291.$$

---

### 4. Оценка точности

Погрешность обратной интерполяции:

$$R = \frac{|g^{(4)}(\eta)|}{4!} \cdot |(y-y_0)(y-y_1)(y-y_2)(y-y_3)|.$$

При $y=0$:

$$|(0 - (-1.000))(0 - (-0.595))(0 - 0.245)(0 - 1.122)| = |1.000 \cdot 0.595 \cdot (-0.245) \cdot (-1.122)|$$

$$= 1.000 \cdot 0.595 \cdot 0.245 \cdot 1.122 \approx 0.1635.$$

Для оценки $|g^{(4)}(\eta)|$ используем связь с производными $f(x)$.  
Примем грубую оценку $|g^{(4)}(\eta)| \sim 0.1$ (учитывая, что $f'(x) \approx 4.2-4.5$ в окрестности корня и производные ограничены).

Таким образом:

$$R \approx \frac{0.1}{24} \cdot 0.1635 \approx 0.00417 \cdot 0.1635 \approx 0.000681.$$

---

Ответ: $x \approx 0.2429$, точность $\approx 6.8 \cdot 10^{-4}$

## Задача VI.9.1

Дана таблица значений функции $f(x) = 4x - \cos x$:

| $x$    | $f(x)$  |
|--------|---------|
| 0.0    | -1.000  |
| 0.1    | -0.595  |
| 0.3    | 0.245   |
| 0.5    | 1.122   |

Требуется найти корень уравнения $4x - \cos x = 0$ методом обратной интерполяции.

---

### 1. Построение обратной интерполяции

Будем искать $x$ как функцию от $y$: $x = g(y)$, где $y = f(x)$.  
В узловых точках:

| $y$    | $x$    |
|--------|--------|
| -1.000 | 0.0    |
| -0.595 | 0.1    |
| 0.245  | 0.3    |
| 1.122  | 0.5    |

Ищем корень при $y = 0$.

---

### 2. Кубическая интерполяция для обратной функции

Используем интерполяционный многочлен Лагранжа степени 3:

$$x = g(y) = \sum_{j=0}^3 x_j \cdot \ell_j(y),$$

где базисные полиномы:

$$\ell_j(y) = \prod_{\substack{k=0 \\ k \ne j}}^3 \frac{y - y_k}{y_j - y_k}.$$

---

**Вычисление базисных полиномов в точке $y=0$:**

Для $j=0$ ($y_0 = -1.000$, $x_0 = 0.0$):

$$\ell_0(0) = \frac{(0 - (-0.595))(0 - 0.245)(0 - 1.122)}{(-1.000 - (-0.595))(-1.000 - 0.245)(-1.000 - 1.122)} = \frac{0.595 \cdot (-0.245) \cdot (-1.122)}{(-0.405) \cdot (-1.245) \cdot (-2.122)}$$

$$= \frac{0.1635}{-1.070} \approx -0.1528.$$

---

Для $j=1$ ($y_1 = -0.595$, $x_1 = 0.1$):

$$\ell_1(0) = \frac{(0 - (-1.000))(0 - 0.245)(0 - 1.122)}{(-0.595 - (-1.000))(-0.595 - 0.245)(-0.595 - 1.122)} = \frac{1.000 \cdot (-0.245) \cdot (-1.122)}{0.405 \cdot (-0.840) \cdot (-1.717)}$$

$$= \frac{0.2749}{0.584} \approx 0.4707.$$

---

Для $j=2$ ($y_2 = 0.245$, $x_2 = 0.3$):

$$\ell_2(0) = \frac{(0 - (-1.000))(0 - (-0.595))(0 - 1.122)}{(0.245 - (-1.000))(0.245 - (-0.595))(0.245 - 1.122)} = \frac{1.000 \cdot 0.595 \cdot (-1.122)}{1.245 \cdot 0.840 \cdot (-0.877)}$$

$$= \frac{-0.6676}{-0.916} \approx 0.7288.$$

---

Для $j=3$ ($y_3 = 1.122$, $x_3 = 0.5$):

$$\ell_3(0) = \frac{(0 - (-1.000))(0 - (-0.595))(0 - 0.245)}{(1.122 - (-1.000))(1.122 - (-0.595))(1.122 - 0.245)} = \frac{1.000 \cdot 0.595 \cdot (-0.245)}{2.122 \cdot 1.717 \cdot 0.877}$$

$$= \frac{-0.1458}{3.194} \approx -0.0456.$$

---

### 3. Вычисление корня

$$x \approx \sum_{j=0}^3 x_j \cdot \ell_j(0) = 0.0 \cdot (-0.1528) + 0.1 \cdot 0.4707 + 0.3 \cdot 0.7288 + 0.5 \cdot (-0.0456)$$

$$= 0 + 0.04707 + 0.21864 - 0.0228 = 0.24291.$$

---

### 4. Оценка точности

Погрешность интерполяции Лагранжа для обратной функции:

$$R = \frac{g^{(4)}(\eta)}{4!} \cdot \prod_{k=0}^3 (y - y_k), \quad \eta \in [\min y_k, \max y_k].$$

При $y = 0$:

$$\prod_{k=0}^3 (0 - y_k) = (0 - (-1.000)) \cdot (0 - (-0.595)) \cdot (0 - 0.245) \cdot (0 - 1.122)$$

$$= 1.000 \cdot 0.595 \cdot (-0.245) \cdot (-1.122) \approx 0.1635.$$

---

**Вычисление четвёртой производной $g^{(4)}(y)$:**

Имеем $y = f(x)$, $x = g(y)$, причём $g'(y) = \frac{1}{f'(x)}$.

Производные обратной функции:

1. $g'(y) = \dfrac{1}{f'(x)}$
2. $g''(y) = -\dfrac{f''(x)}{[f'(x)]^3}$
3. $g'''(y) = \dfrac{3[f''(x)]^2 - f'(x)f'''(x)}{[f'(x)]^5}$
4. $g^{(4)}(y) = -\dfrac{15[f''(x)]^3 - 10f'(x)f''(x)f'''(x) + [f'(x)]^2 f^{(4)}(x)}{[f'(x)]^7}$

Для $f(x) = 4x - \cos x$:

$$
\begin{aligned}
f'(x) &= 4 + \sin x \\
f''(x) &= \cos x \\
f'''(x) &= -\sin x \\
f^{(4)}(x) &= -\cos x
\end{aligned}
$$

В окрестности корня $x^* \approx 0.2429$:

$$
\begin{aligned}
f'(x^*) &\approx 4 + \sin(0.2429) \approx 4.240 \\
f''(x^*) &\approx \cos(0.2429) \approx 0.970 \\
f'''(x^*) &\approx -\sin(0.2429) \approx -0.240 \\
f^{(4)}(x^*) &\approx -\cos(0.2429) \approx -0.970
\end{aligned}
$$

Подставляем в формулу для $g^{(4)}$:

$$
\begin{aligned}
g^{(4)}(y^*) &\approx -\frac{15 \cdot (0.970)^3 - 10 \cdot 4.240 \cdot 0.970 \cdot (-0.240) + (4.240)^2 \cdot (-0.970)}{(4.240)^7} \\
&= -\frac{15 \cdot 0.9127 - 10 \cdot 4.240 \cdot (-0.2328) + 17.9776 \cdot (-0.970)}{4.240^7} \\
&= -\frac{13.6905 + 9.866 - 17.438}{4.240^7} \\
&= -\frac{6.1185}{4.240^7}
\end{aligned}
$$

Вычислим $4.240^7$:

$$
4.240^2 = 17.9776, \quad 4.240^4 = (17.9776)^2 \approx 323.2, \quad 4.240^7 = 323.2 \cdot 4.240^3 \approx 323.2 \cdot 76.2 \approx 24600
$$

Тогда:

$$|g^{(4)}(\eta)| \approx \frac{6.12}{24600} \approx 2.49 \cdot 10^{-4}.$$

---

**Окончательная оценка погрешности:**

$$R \approx \frac{|g^{(4)}(\eta)|}{24} \cdot 0.1635 \approx \frac{2.49 \cdot 10^{-4}}{24} \cdot 0.1635 \approx 1.04 \cdot 10^{-5} \cdot 0.1635 \approx 1.70 \cdot 10^{-6}.$$

---

### 5. Ответ

Корень уравнения: $x^* \approx 0.2429$

Погрешность метода обратной интерполяции: $\Delta x \approx 1.7 \cdot 10^{-6}$

Учитывая возможные ошибки округления при вычислениях, окончательно:

$$\boxed{x = 0.2429 \pm 0.0001}$$

### 6. Проверка

Подстановка в исходное уравнение:

$$4 \cdot 0.2429 - \cos(0.2429) \approx 0.9716 - 0.9708 \approx 0.0008,$$

что согласуется с найденной точностью.

Константы Лебега

Для интерполяции Лагранжа константа Лебега определяется как:

$$\Lambda_n = \max_{y \in [a,b]} \sum_{j=0}^n |\ell_j(y)|$$

где $\ell_j(y)$ — базисные полиномы Лагранжа.

$$|\Delta P(y)| \leq \Lambda_n \cdot \max_j |\delta x_j|$$

В точке $y=0$:

$$
\begin{aligned}
\Lambda_3(0) &= |\ell_0(0)| + |\ell_1(0)| + |\ell_2(0)| + |\ell_3(0)| \\
&= 0.152803 + 0.470489 + 0.728361 + 0.045634 \\
&= 1.397287
\end{aligned}
$$


Табличные $x_j$ даны с погрешностью $\delta x = 0.0005$  
Тогда максимальная погрешность интерполяции:

$$\Delta_{\text{данные}} \leq \Lambda_3(0) \cdot \delta x = 1.397287 \cdot 0.0005 \approx 6.99 \times 10^{-4}$$

$$
\begin{aligned}

## Задача 1.6

Дана функция $g(x) = \cosh 3x$ на отрезке $[1, 3]$.

### 1. Аналитическая оценка погрешности интерполяции

Погрешность интерполяционного многочлена Лагранжа степени $n$:
$$R_n(x) = \frac{g^{(n+1)}(\xi)}{(n+1)!} \cdot \omega_{n+1}(x),$$
где $\omega_{n+1}(x) = (x-x_0)(x-x_1)\dots(x-x_n)$, $\xi \in [1,3]$.

#### 1.1 Производные функции

$g(x) = \cosh 3x = \frac{e^{3x} + e^{-3x}}{2}$

$g^{(k)}(x) = 3^k \cdot \cosh 3x$ при чётных $k$,
$g^{(k)}(x) = 3^k \cdot \sinh 3x$ при нечётных $k$

На отрезке $[1,3]$: $\max|\cosh 3x| = \cosh 9$, $\max|\sinh 3x| = \sinh 9$

Таким образом: $M_{n+1} = \max\limits_{x\in[1,3]}|g^{(n+1)}(x)| = 3^{n+1} \cdot \max(\cosh 9, \sinh 9) = 3^{n+1} \cdot \sinh 9$

#### 1.2 Равномерная сетка

$x_i = 1 + i \cdot h$, где $h = \frac{2}{n}$, $i = 0,1,\dots,n$

Максимум $|\omega_{n+1}(x)|$ на $[1,3]$ оценивается как:
$$\max|\omega_{n+1}(x)| \leq \frac{h^{n+1}}{4} \cdot (n+1)!$$

Погрешность:
$$|R_n(x)| \leq \frac{3^{n+1} \cdot \sinh 9}{(n+1)!} \cdot \frac{h^{n+1}}{4} \cdot (n+1)! = \frac{3^{n+1} \cdot \sinh 9 \cdot h^{n+1}}{4}$$

При $h = \frac{2}{n}$:
$$|R_n| \leq \frac{3^{n+1} \cdot \sinh 9 \cdot (2/n)^{n+1}}{4} = \frac{3^{n+1} \cdot 2^{n+1} \cdot \sinh 9}{4 \cdot n^{n+1}}$$

#### 1.3 Чебышевская сетка

Узлы Чебышева на $[1,3]$:
$$x_i = 2 + \cos\left(\frac{(2i+1)\pi}{2(n+1)}\right), \quad i=0,1,\dots,n$$

Для чебышевской сетки:
$$\max|\omega_{n+1}(x)| \leq \frac{(b-a)^{n+1}}{2^{2n+1}} = \frac{2^{n+1}}{2^{2n+1}} = \frac{1}{2^n}$$

Погрешность:
$$|R_n(x)| \leq \frac{3^{n+1} \cdot \sinh 9}{(n+1)!} \cdot \frac{1}{2^n}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial, cosh, sinh

def g(x):
    return np.cosh(3*x)

def lagrange_interpolation(x, x_nodes, y_nodes):
    n = len(x_nodes)
    result = 0.0
    for i in range(n):
        term = y_nodes[i]
        for j in range(n):
            if i != j:
                term *= (x - x_nodes[j]) / (x_nodes[i] - x_nodes[j])
        result += term
    return result

def uniform_nodes(a, b, n):
    return np.linspace(a, b, n+1)

def chebyshev_nodes(a, b, n):
    nodes = []
    for i in range(n+1):
        cos_val = np.cos((2*i + 1) * np.pi / (2*(n+1)))
        nodes.append(0.5*(a + b) + 0.5*(b - a)*cos_val)
    return np.array(nodes)

a, b = 1, 3
test_points = np.linspace(a, b, 1000)
exact_values_double = g(test_points)
exact_values_single = exact_values_double.astype(np.float32)

sinh9 = sinh(9)
n_values = np.arange(1, 21)

uniform_analytic = []
chebyshev_analytic = []

for n in n_values:
    if n <= 15:
        log_R_uniform = (n+1)*np.log(6.0) + np.log(sinh9) - np.log(4.0) - (n+1)*np.log(float(n))
        R_uniform = np.exp(log_R_uniform)
    else:
        R_uniform = np.nan
    uniform_analytic.append(R_uniform)

    log_R_chebyshev = (n+1)*np.log(3.0) + np.log(sinh9) - np.log(float(factorial(n+1))) - n*np.log(2.0)
    R_chebyshev = np.exp(log_R_chebyshev)
    chebyshev_analytic.append(R_chebyshev)

uniform_errors_double = []
chebyshev_errors_double = []

uniform_errors_single = []
chebyshev_errors_single = []

for n in n_values:
    x_uniform = uniform_nodes(a, b, n)
    y_uniform_double = g(x_uniform)
    y_uniform_single = y_uniform_double.astype(np.float32)

    x_chebyshev = chebyshev_nodes(a, b, n)
    y_chebyshev_double = g(x_chebyshev)
    y_chebyshev_single = y_chebyshev_double.astype(np.float32)

    interp_uniform_double = np.array([lagrange_interpolation(x, x_uniform, y_uniform_double) for x in test_points])
    interp_chebyshev_double = np.array([lagrange_interpolation(x, x_chebyshev, y_chebyshev_double) for x in test_points])

    error_uniform_double = np.max(np.abs(interp_uniform_double - exact_values_double))
    error_chebyshev_double = np.max(np.abs(interp_chebyshev_double - exact_values_double))

    uniform_errors_double.append(error_uniform_double)
    chebyshev_errors_double.append(error_chebyshev_double)

    interp_uniform_single = np.array([lagrange_interpolation(x, x_uniform, y_uniform_single) for x in test_points], dtype=np.float32)
    interp_chebyshev_single = np.array([lagrange_interpolation(x, x_chebyshev, y_chebyshev_single) for x in test_points], dtype=np.float32)

    error_uniform_single = np.max(np.abs(interp_uniform_single - exact_values_single))
    error_chebyshev_single = np.max(np.abs(interp_chebyshev_single - exact_values_single))

    uniform_errors_single.append(error_uniform_single)
    chebyshev_errors_single.append(error_chebyshev_single)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

n_valid = n_values[np.array(uniform_analytic) <= 1e10]
uniform_valid = [x for x in uniform_analytic if x <= 1e10]

ax1.semilogy(n_valid, uniform_valid, 'b-', label='Аналитическая оценка (равномерная)')
ax1.semilogy(n_values, chebyshev_analytic, 'r-', label='Аналитическая оценка (Чебышев)')
ax1.semilogy(n_values, uniform_errors_double, 'bo--', label='Численная ошибка (равномерная)')
ax1.semilogy(n_values, chebyshev_errors_double, 'ro--', label='Численная ошибка (Чебышев)')
ax1.set_xlabel('Порядок интерполяции n')
ax1.set_ylabel('Погрешность')
ax1.set_title('Двойная точность')
ax1.legend()
ax1.grid(True)

ax2.semilogy(n_valid, uniform_valid, 'b-', label='Аналитическая оценка (равномерная)')
ax2.semilogy(n_values, chebyshev_analytic, 'r-', label='Аналитическая оценка (Чебышев)')
ax2.semilogy(n_values, uniform_errors_single, 'bo--', label='Численная ошибка (равномерная)')
ax2.semilogy(n_values, chebyshev_errors_single, 'ro--', label='Численная ошибка (Чебышев)')
ax2.set_xlabel('Порядок интерполяции n')
ax2.set_ylabel('Погрешность')
ax2.set_title('Одинарная точность')
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()

print("Минимальные погрешности (двойная точность):")
print(f"Равномерная сетка: {min(uniform_errors_double):.2e}")
print(f"Чебышевская сетка: {min(chebyshev_errors_double):.2e}")

print("\nМинимальные погрешности (одинарная точность):")
print(f"Равномерная сетка: {min(uniform_errors_single):.2e}")
print(f"Чебышевская сетка: {min(chebyshev_errors_single):.2e}")
```

## Выводы

1. Теоретические оценки верны, но для равномерной сетки дают завышенные значения погрешности

2. Чебышевская сетка существенно превосходит равномерную, так как монотонно сходится при увеличении n, а также возможно достижение машинной точности

3. Равномерная сетка проявляет неустойчивость: рост погрешности при n > 15
