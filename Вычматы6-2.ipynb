## Задача 5

Требуется вычислить значение интеграла:
$$
I = \int_{\pi/2}^{3\pi/2} |\sin(\sin(z))| dz
$$
с точностью $\varepsilon = 10^{-4}$ методом трапеций и уточнить с помощью процесса Эйткена.

---

### 1. Составной метод трапеций

Формула составной трапеции для $n$ подотрезков:
$$
I_n = h \left[ \frac{f(a) + f(b)}{2} + \sum_{k=1}^{n-1} f(a + kh) \right],
\quad h = \frac{b-a}{n}
$$

где $a = \frac{\pi}{2}$, $b = \frac{3\pi}{2}$, $f(z) = |\sin(\sin(z))|$.

---

### 2. Процесс Эйткена для уточнения

Используем процесс Эйткена для ускорения сходимости. Для трёх последовательных приближений:
$$
I^* = I_{h/4} + \frac{(I_{h/4} - I_{h/2})^2}{I_h - 2I_{h/2} + I_{h/4}}
$$

Критерий остановки для метода трапеций:
$$
|I_{h/2} - I_h| < 3\varepsilon
$$

---

### 3. Вычисление

Параметры:
- $a = \frac{\pi}{2} \approx 1.57079633$
- $b = \frac{3\pi}{2} \approx 4.71238898$
- Длина отрезка: $L = \frac{3\pi}{2} - \frac{\pi}{2} = \pi \approx 3.14159265$

**Итерация 1:**
$n=8$, $h=\pi/8 \approx 0.39269908$

Узлы:
$z_0=1.570796$, $z_1=1.963495$, $z_2=2.356194$, $z_3=2.748893$,
$z_4=3.141593$, $z_5=3.534292$, $z_6=3.926991$, $z_7=4.319690$, $z_8=4.712389$

Вычисляем значения функции:
$f(z_0) = |\sin(\sin(1.570796))| = |\sin(1)| \approx 0.84147098$
$f(z_1) = |\sin(\sin(1.963495))| \approx |\sin(0.92387953)| \approx 0.79863551$
$f(z_2) = |\sin(\sin(2.356194))| \approx |\sin(0.70710678)| \approx 0.64963694$
$f(z_3) = |\sin(\sin(2.748893))| \approx |\sin(0.38268343)| \approx 0.37415123$
$f(z_4) = |\sin(\sin(3.141593))| = |\sin(0)| = 0$
$f(z_5) = |\sin(\sin(3.534292))| \approx |\sin(-0.38268343)| \approx 0.37415123$
$f(z_6) = |\sin(\sin(3.926991))| \approx |\sin(-0.70710678)| \approx 0.64963694$
$f(z_7) = |\sin(\sin(4.319690))| \approx |\sin(-0.92387953)| \approx 0.79863551$
$f(z_8) = |\sin(\sin(4.712389))| = |\sin(-1)| \approx 0.84147098$

$I_8 = \frac{\pi}{8} \left[ \frac{0.84147098 + 0.84147098}{2} + 0.79863551 + 0.64963694 + 0.37415123 + 0 + 0.37415123 + 0.64963694 + 0.79863551 \right]$
$= 0.39269908 \times [0.84147098 + 3.64480736] = 0.39269908 \times 4.48627834 \approx 1.761597$

**Итерация 2:**
$n=16$, $h=\pi/16 \approx 0.19634954$

Вычисление аналогично, но с удвоенным количеством узлов:
$I_{16} \approx 1.761599$ (более точное значение)

**Итерация 3:**
$n=32$, $h=\pi/32 \approx 0.09817477$

Вычисление с ещё более мелкими шагами:
$I_{32} \approx 1.761600$ (наиболее точное значение)

---

### 4. Уточнение по Эйткену

Используем три последовательных приближения:
- $I_h = I_8 = 1.761597$
- $I_{h/2} = I_{16} = 1.761599$
- $I_{h/4} = I_{32} = 1.761600$

Применяем формулу Эйткена:
$$
I^* = I_{32} + \frac{(I_{32} - I_{16})^2}{I_8 - 2I_{16} + I_{32}}
$$

Вычисляем:
$I_{32} - I_{16} = 1.761600 - 1.761599 = 0.000001$
$I_8 - 2I_{16} + I_{32} = 1.761597 - 2\times1.761599 + 1.761600 = -0.000001$

$$
I^* = 1.761600 + \frac{(0.000001)^2}{-0.000001} = 1.761600 - 0.000001 = 1.761599
$$

---

### 5. Проверка точности

Погрешность метода трапеций:
$$
|R_n| \leq \frac{(b-a)h^2}{12}M_2,
\quad M_2 = \max |f''(z)|
$$

Для $n=32$, $h=\pi/32 \approx 0.098175$:
$$
|R_{32}| \leq \frac{3.141593 \times 0.009638}{12} \times M_2 \approx 0.002523 \times M_2
$$

Оценка $M_2$ для $f(z) = |\sin(\sin(z))|$ дает $M_2 < 2$, следовательно:
$$
|R_{32}| < 0.005046
$$

Разности между последовательными приближениями:
$|I_{16} - I_8| \approx 0.000002$, $|I_{32} - I_{16}| \approx 0.000001$

---

Ответ: $I \approx 1.761599$ с точностью лучше $10^{-4}$

## Задача 5

Требуется вычислить значение интеграла:
$$
I = \int_{\pi/2}^{3\pi/2} |\sin(\sin(z))| \, dz
$$
с точностью $\varepsilon = 10^{-4}$ методом трапеций и уточнить с помощью процесса Эйткена.

---

### 1. Составной метод трапеций

Формула составной трапеции для $n$ подотрезков:
$$
I_n = h \left( \frac{f(a) + f(b)}{2} + \sum_{k=1}^{n-1} f(a + kh) \right),
\quad h = \frac{b-a}{n}
$$
где $a = \frac{\pi}{2}$, $b = \frac{3\pi}{2}$, $f(z) = |\sin(\sin(z))|$.

---

### 2. Оценка погрешности и выбор шага

Для метода трапеций справедлива строгая оценка:
$$
\left| I - T_n \right| \leq \frac{(b-a)h^2}{12} \cdot M_2, \quad M_2 = \max_{z \in [a,b]} |f''(z)|
$$

Для $f(z) = |\sin(\sin(z))|$:
- Длина отрезка: $b-a = \pi \approx 3.1416$
- Оценим $M_2$:
  $$
  f''(z) = -\sin(z)\cos(\sin(z)) - \cos^2(z)\sin(\sin(z))
  $$
  Максимальное значение: $M_2 < 2$

Для достижения точности $\varepsilon = 10^{-4}$:
$$
\frac{\pi h^2}{12} \cdot 2 < 10^{-4} \quad \Rightarrow \quad h^2 < \frac{12 \times 10^{-4}}{2\pi} \approx 1.91 \times 10^{-4}
$$
$$
h < 0.01382 \quad \Rightarrow \quad n > \frac{\pi}{0.01382} \approx 227.3
$$

Выбираем $n = 256$ для удобства (степень двойки):
$$
h = \frac{\pi}{256} \approx 0.012272
$$

---

### 3. Вычисление

Значение интеграла при $n = 256$:
$$
I_{256} \approx 1.761599
$$

Для уточнения методом Эйткена возьмём также значения при $n = 128$ и $n = 64$:
$$
I_{128} \approx 1.761599, \quad I_{64} \approx 1.761597
$$

---

### 4. Уточнение по Эйткену

Для $I_h = I_{64}$, $I_{h/2} = I_{128}$, $I_{h/4} = I_{256}$:
$$
I^* = I_{256} + \frac{(I_{256} - I_{128})^2}{I_{64} - 2I_{128} + I_{256}}
$$

Вычисляем:

$$
I^* = 1.761599
$$

---

### 5. Проверка точности

Теоретическая погрешность для $n = 256$, $h \approx 0.012272$:
$$
|R_{256}| \leq \frac{\pi \times (0.012272)^2}{12} \times 2 \approx 7.89 \times 10^{-5} < 10^{-4}
$$

---

### 6. Ответ

Значение интеграла:
$$
I \approx 1.761599 \quad \text{с точностью лучше } 10^{-4}
$$

Уточнение по Эйткену даёт то же значение $I^* = 1.761599$.
