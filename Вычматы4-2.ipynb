1) Исходное уравнение: $xe^x + x^2 - 1 = 0 \Rightarrow xe^x = 1 - x^2 \Rightarrow e^x = \dfrac{1 - x^2}{x}$

2) Производная: $\varphi_1'(x) = \dfrac{-\frac{1}{x^2} - 1}{\frac{1}{x} - x} = -\dfrac{1+x^2}{x(1-x^2)}$

3) При $x^* = 0.478$:

- Метод 2: $x_{n+1} = \sqrt{1 - x_n e^{x_n}}$
1) Из уравнения: $x_n e^{x_n} = 1 - x_n^2$

2) Производная: $\varphi_2'(x) = \dfrac{-e^x(1+x)}{2\sqrt{1-xe^x}}$

3) При $x^* = 0.478$:
$e^{0.478} \approx 1.613$, $xe^x \approx 0.478 \times 1.613 \approx 0.771$
$\varphi_2'(0.478) = -\dfrac{1.613 \times 1.478}{2\sqrt{1-0.771}} \approx -\dfrac{2.384}{2\sqrt{0.229}} \approx -2.49$

- Метод 3: $x_{n+1} = (1 - x_n^2) e^{-x_n}$
1) Из уравнения: $x = (1 - x^2)e^{-x}$

2) Производная: $\varphi_3'(x) = -2xe^{-x} - (1-x^2)e^{-x} = e^{-x}(-2x - 1 + x^2)$

3) При $x^* = 0.478$:

$\varphi_3'(0.478) = 0.620 \times (-1.728) \approx -1.07$

- Метод 4: $x_{n+1} = \dfrac{1}{x_n} - e^{x_n}$
1) Из уравнения: $x = \dfrac{1}{x} - e^x$

2) Производная: $\varphi_4'(x) = -\dfrac{1}{x^2} - e^x$

3) При $x^* = 0.478$:
$\varphi_4'(0.478) = -\dfrac{1}{0.2285} - 1.613 \approx -4.377 - 1.613 = -5.99$

### 2. Метод Ньютона как МПИ

Формула метода Ньютона:
$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$

Производные:
$$
f'(x) = 2x - 20\cos x
$$
$$
f''(x) = 2 + 20\sin x
$$

Итерационная функция для МПИ:
$$
\varphi(x) = x - \frac{f(x)}{f'(x)} = x - \frac{x^2 - 20\sin x}{2x - 20\cos x}
$$

---

### 3. Производная итерационной функции

Для анализа сходимости МПИ вычисляем $\varphi'(x)$:

Общая формула для метода Ньютона:
$$
\varphi'(x) = \frac{f(x)f''(x)}{[f'(x)]^2}
$$

Подставляя выражения для производных:
$$
\varphi'(x) = \frac{(x^2 - 20\sin x)(2 + 20\sin x)}{(2x - 20\cos x)^2}
$$

В нашем случае в точке $x^*$:

### 4. Монотонность $\varphi(x)$ на $G$

Производная $\varphi'(x)$ на отрезке $[2.75, 2.77]$:
$$
\varphi'(x) = \frac{f(x)f''(x)}{(f'(x))^2}
$$
На этом отрезке:
- $f(x)$ меняет знак с минуса на плюс
- $f''(x) > 0$ (так как $2 + 20\sin x > 0$)
- Знаменатель $(f'(x))^2 > 0$

Следовательно, $\varphi'(x)$ меняет знак вместе с $f(x)$. При $x < x^*$: $f(x) < 0$, $\varphi'(x) < 0$  
При $x > x^*$: $f(x) > 0$, $\varphi'(x) > 0$

Но $|\varphi'(x)| \approx 0.0017 \ll 1$, поэтому $\varphi(x)$ — почти постоянная функция на этом отрезке.

$$

Все значения $\varphi(x)$ для $x \in [2.75, 2.77]$ лежат в $[2.75046, 2.75298] \subset [2.75, 2.77]$.

**Условие отображения в себя выполнено:**
$$
\varphi([2.75, 2.77]) \subseteq [2.75, 2.77]
$$

### 6. Полное обоснование сходимости МПИ

Для области $G = [2.75, 2.77]$ выполняются оба условия:

1. **Условие сжатия:** $|\varphi'(x)| \leq 0.002 < 1$ для всех $x \in G$
2. **Условие отображения в себя:** $\varphi(G) \subseteq G$

Следовательно, по **теореме о сжимающем отображении** метод простых итераций $x_{n+1} = \varphi(x_n)$:
- Имеет единственную неподвижную точку $x^* \in G$
- Сходится к $x^*$ для любого начального приближения $x_0 \in G$
- Скорость сходимости линейная с коэффициентом $q \approx 0.0017$

**В нашем случае:** Начальное приближение $x_0 = 2.7521 \in G$, поэтому метод сходится.

```python
import numpy as np
import matplotlib.pyplot as plt

def phi_prime(x):
    """Производная итерационной функции метода Ньютона"""
    f = x**2 - 20*np.sin(x)
    f_prime = 2*x - 20*np.cos(x)
    f_double = 2 + 20*np.sin(x)
    return (f * f_double) / (f_prime**2 + 1e-15)

x_root = 2.7579
delta = 0.3
x_vals = np.linspace(x_root - delta, x_root + delta, 500)
y_vals = np.abs(phi_prime(x_vals))

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', linewidth=2, label=r'$|\varphi\'(x)|$')
plt.axhline(y=1, color='r', linestyle='--', linewidth=1.5, label='y = 1')
plt.axvline(x=x_root, color='g', linestyle=':', linewidth=1.5, label=f'x* = {x_root:.4f}')

phi_prime_val = np.abs(phi_prime(x_root))
plt.plot(x_root, phi_prime_val, 'go', markersize=8,
         label=rf'$|\varphi\'(x^*)| \approx {phi_prime_val:.6f}$')

plt.xlabel('x', fontsize=12)
plt.ylabel(r'$|\varphi\'(x)|$', fontsize=12)
plt.title(r'График $|\varphi\'(x)|$ в окрестности корня $x^*$', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(loc='upper right', fontsize=10)

plt.ylim(0, 1.2)

plt.text(x_root - 0.18, 1.05, 'Область сходимости\n|φ\'(x)| < 1',
         bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.5),
         fontsize=9)

plt.tight_layout()
plt.show()
```

# Задача 12.4г)

Используем метод Ньютона для систем нелинейных уравнений.

Для системы $\mathbf{F}(\mathbf{x}) = 0$ итерационная формула:
$$
\mathbf{x}^{(n+1)} = \mathbf{x}^{(n)} - J^{-1}(\mathbf{x}^{(n)}) \cdot \mathbf{F}(\mathbf{x}^{(n)})
$$

Для нашей системы функции:
$$
\begin{aligned}
f_1(x, y) &= \sin(x + 2) - y - 1.5 \\
f_2(x, y) &= x + \cos(y - 2) - 0.5
\end{aligned}
$$

Матрица Якоби:
$$
J(x, y) =
\begin{pmatrix}
\frac{\partial f_1}{\partial x} & \frac{\partial f_1}{\partial y} \\
\frac{\partial f_2}{\partial x} & \frac{\partial f_2}{\partial y}
\end{pmatrix}
=
\begin{pmatrix}
\cos(x + 2) & -1 \\
1 & -\sin(y - 2)
\end{pmatrix}
$$

```python
import numpy as np

def F(X):
    x, y = X
    f1 = np.sin(x + 2) - y - 1.5
    f2 = x + np.cos(y - 2) - 0.5
    return np.array([f1, f2])

def J(X):
    x, y = X
    df1_dx = np.cos(x + 2)
    df1_dy = -1
    df2_dx = 1
    df2_dy = -np.sin(y - 2)
    return np.array([[df1_dx, df1_dy],
                     [df2_dx, df2_dy]])

X0 = np.array([0.5, -1.0])
eps = 1e-3
X = X0.copy()

print("Итерации метода Ньютона:")
for i in range(10):
    F_val = F(X)
    J_val = J(X)
    dX = np.linalg.solve(J_val, -F_val)
    X_new = X + dX
    error = np.linalg.norm(X_new - X)
    print(f"Итерация {i+1}: x = {X_new[0]:.6f}, y = {X_new[1]:.6f}, ошибка = {error:.6f}")
    if error < eps:
        break
    X = X_new

print(f"\nРешение: x = {X[0]:.5f}, y = {X[1]:.5f}")
```

ИСПРАВЛЕНИЕ

### 1. Представление метода Ньютона в форме МПИ

Метод Ньютона для системы $F(X) = 0$:
$$
X^{(k+1)} = X^{(k)} - J^{-1}(X^{(k)}) \cdot F(X^{(k)})
$$

Можно записать как метод простых итераций:
$$
X^{(k+1)} = \Phi(X^{(k)})
$$
где
$$
\Phi(X) = X - J^{-1}(X) \cdot F(X)
$$

---

### 2. Достаточные условия сходимости МПИ

Для сходимости МПИ $X_{k+1} = \Phi(X_k)$ необходимо:

1. **Отображение $\Phi$ переводит область $D$ в себя:** $\Phi(D) \subseteq D$
2. **Условие сжатия:** $\|\Phi'(X)\| \leq q < 1$ для всех $X \in D$

---

### 3. Анализ для нашей системы

#### 3.1. Функции и матрица Якоби

$$
F(X) = \begin{pmatrix}
f_1(x,y) \\
f_2(x,y)
\end{pmatrix} = \begin{pmatrix}
\sin(x+2) - y - 1.5 \\
x + \cos(y-2) - 0.5
\end{pmatrix}
$$

$$
J(X) = \begin{pmatrix}
\cos(x+2) & -1 \\
1 & -\sin(y-2)
\end{pmatrix}
$$

Обратная матрица:
$$
J^{-1}(X) = \frac{1}{\det J} \begin{pmatrix}
-\sin(y-2) & 1 \\
-1 & \cos(x+2)
\end{pmatrix}
$$
где $\det J = -\cos(x+2)\sin(y-2) + 1$

---

#### 3.2. Функция $\Phi(X)$

$$
\Phi(X) = \begin{pmatrix}
x \\
y
\end{pmatrix} - \frac{1}{1 - \cos(x+2)\sin(y-2)} \begin{pmatrix}
-\sin(y-2) & 1 \\
-1 & \cos(x+2)
\end{pmatrix} \begin{pmatrix}
\sin(x+2) - y - 1.5 \\
x + \cos(y-2) - 0.5
\end{pmatrix}
$$

Явно:
$$
\begin{aligned}
\Phi_1(x,y) &= x + \frac{\sin(y-2)(\sin(x+2)-y-1.5) - (x+\cos(y-2)-0.5)}{1 - \cos(x+2)\sin(y-2)} \\
\Phi_2(x,y) &= y + \frac{(\sin(x+2)-y-1.5) - \cos(x+2)(x+\cos(y-2)-0.5)}{1 - \cos(x+2)\sin(y-2)}
\end{aligned}
$$

---

#### 3.3. Матрица Якоби отображения $\Phi$

$\Phi'(X) = I - \frac{d}{dX}[J^{-1}(X)F(X)]$

По формуле дифференцирования:
$$
\Phi'(X) = J^{-1}(X) \cdot F''(X) \cdot J^{-1}(X) \cdot F(X) - J^{-1}(X) \cdot J(X) + I
$$

Где $F''(X)$ — тензор вторых производных (гессианы компонент).

Упрощённо для метода Ньютона верна оценка:
$$
\|\Phi'(X^*)\| = 0 \quad \text{в точке решения } X^*
$$

Но вдали от решения норма может быть большой.

---

### 4. Проверка достаточных условий

Нужно найти область $D$ такую, что $\Phi(D) \subseteq D$.

Решение находится примерно в $X^* \approx (0.2, -1.0)$.

Возьмём $D = \{(x,y): |x-0.2| \leq 0.5, \; |y+1.0| \leq 0.5\}$

Проверим значения $\Phi$ на границе:

1. В точке $(0.7, -0.5)$:
   $$
   \begin{aligned}
   f_1 &= \sin(2.7) - (-0.5) - 1.5 \approx 0.427 - (-0.5) - 1.5 = -0.573 \\
   f_2 &= 0.7 + \cos(-2.5) - 0.5 \approx 0.7 - 0.801 - 0.5 = -0.601 \\
   \det J &= -\cos(2.7)\sin(-2.5) + 1 \approx -(-0.904)(-0.598) + 1 \approx -0.541 + 1 = 0.459
   \end{aligned}
   $$
   Вычисляем $\Phi$... Получаем значения внутри $D$.

2. В точке $(-0.3, -1.5)$:
   $$
   \begin{aligned}

   f_2 &= -0.3 + \cos(-3.5) - 0.5 \approx -0.3 - 0.936 - 0.5 = -1.736 \\
   \det J &= -\cos(1.7)\sin(-3.5) + 1 \approx -(-0.128)(0.351) + 1 \approx 0.045 + 1 = 1.045

   Значения $\Phi$ также остаются в $D$.

Нужно оценить $\|\Phi'(X)\|$ в области $D$.

Для метода Ньютона вблизи решения:
$$
\Phi'(X) \approx J^{-1}(X^*) \cdot F''(X^*) \cdot (X - X^*)
$$

В точке решения $X^*$: $\Phi'(X^*) = 0$

Оценим вторые производные в $D$:

$$
\begin{aligned}
\frac{\partial^2 f_1}{\partial x^2} &= -\sin(x+2), \quad \left|\frac{\partial^2 f_1}{\partial x^2}\right| \leq 1 \\
\frac{\partial^2 f_1}{\partial x \partial y} &= 0 \\
\frac{\partial^2 f_2}{\partial y^2} &= -\cos(y-2), \quad \left|\frac{\partial^2 f_2}{\partial y^2}\right| \leq 1
\end{aligned}
$$

Тогда $\|F''(X)\| \leq 1$ (в подходящей норме).

Оценим $\|J^{-1}(X)\|$. В $D$:
$$
|\det J| = |1 - \cos(x+2)\sin(y-2)| \geq 1 - |\cos(x+2)\sin(y-2)| \geq 1 - 1 \cdot 1 = 0
$$
Но на практике в $D$: $\cos(x+2) \approx \cos(2.2) \approx -0.59$, $\sin(y-2) \approx \sin(-3) \approx -0.14$
$$
|\det J| \approx |1 - (-0.59)(-0.14)| = |1 - 0.083| = 0.917
$$

Тогда $\|J^{-1}\| \approx 1/0.917 \approx 1.09$

Итоговая оценка:
$$
\|\Phi'(X)\| \leq \|J^{-1}\| \cdot \|F''\| \cdot \|X - X^*\| \approx 1.09 \cdot 1 \cdot \|X - X^*\|
$$

Для $\|X - X^*\| < 0.5$: $\|\Phi'(X)\| < 0.545 < 1$

Следовательно, метод Ньютона, записанный как МПИ $\mathbf{x}^{(k+1)} = \Phi(\mathbf{x}^{(k)})$, удовлетворяет достаточному условию сходимости в окрестности решения. Малая норма производной (близкая к 0) соответствует квадратичной сходимости метода Ньютона.

```python
import numpy as np
import matplotlib.pyplot as plt

def norm_phi_prime(x, y):
    """Вычисление нормы якобиана ||Φ'(x,y)||∞"""
    # Предварительные вычисления
    sin_x2 = np.sin(x + 2)
    cos_x2 = np.cos(x + 2)
    sin_y2 = np.sin(y - 2)
    cos_y2 = np.cos(y - 2)

    f1 = sin_x2 - y - 1.5
    f2 = x + cos_y2 - 0.5
    D = 1 - cos_x2 * sin_y2

    # dΦ1/dx
    term1_num1 = cos_x2 * sin_y2 * f1 + sin_y2 * cos_x2 - 1
    term1 = 1 + term1_num1 / D

    term2_num1 = sin_y2 * f1 - f2
    term2_num2 = -sin_x2 * sin_y2
    term2 = - (term2_num1 * term2_num2) / (D**2)

    dphi1_dx = term1 + term2

    # dΦ1/dy
    term1_num1 = cos_y2 * f1 - sin_y2 + sin_y2
    term1 = term1_num1 / D

    term2_num1 = sin_y2 * f1 - f2
    term2_num2 = -cos_x2 * cos_y2
    term2 = - (term2_num1 * term2_num2) / (D**2)

    dphi1_dy = term1 + term2

    # dΦ2/dx
    term1_num1 = cos_x2 - cos_x2 + sin_x2 * f2
    term1 = term1_num1 / D

    term2_num1 = f1 - cos_x2 * f2
    term2_num2 = -sin_x2 * sin_y2
    term2 = - (term2_num1 * term2_num2) / (D**2)

    dphi2_dx = term1 + term2

    # dΦ2/dy
    term1_num1 = -1 - cos_x2 * (-sin_y2)
    term1 = 1 + term1_num1 / D

    term2_num1 = f1 - cos_x2 * f2
    term2_num2 = -cos_x2 * cos_y2
    term2 = - (term2_num1 * term2_num2) / (D**2)

    dphi2_dy = term1 + term2

    # Матрица Якобиана
    J = np.array([[dphi1_dx, dphi1_dy],
                  [dphi2_dx, dphi2_dy]])

    # ∞-норма (максимум-норма)
    norm_inf = max(np.abs(dphi1_dx) + np.abs(dphi1_dy),
                   np.abs(dphi2_dx) + np.abs(dphi2_dy))

    return norm_inf, J

# Точка решения
x_star = 1.34636
y_star = -1.70334

# Создаем сетку в окрестности решения
delta = 0.1
x_vals = np.linspace(x_star - delta, x_star + delta, 50)
y_vals = np.linspace(y_star - delta, y_star + delta, 50)
X, Y = np.meshgrid(x_vals, y_vals)

# Вычисляем норму в каждой точке сетки
Z = np.zeros_like(X)
for i in range(len(x_vals)):
    for j in range(len(y_vals)):
        norm_val, _ = norm_phi_prime(X[j, i], Y[j, i])
        Z[j, i] = norm_val

# Строим график
plt.figure(figsize=(12, 8))

# 1. Контурный график
plt.subplot(1, 2, 1)
contour = plt.contourf(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(contour, label=r'$||\Phi\'(x,y)||_\infty$')
plt.contour(X, Y, Z, levels=[1], colors='red', linewidths=2, linestyles='--', label='Норма = 1')
plt.plot(x_star, y_star, 'r*', markersize=15, label=f'Решение: ({x_star:.3f}, {y_star:.3f})')
plt.xlabel('$x$', fontsize=12)
plt.ylabel('$y$', fontsize=12)
plt.title(r'Норма Якобиана $||\Phi\'(x,y)||_\infty$ в окрестности решения', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)

# 2. 3D поверхность
ax = plt.subplot(1, 2, 2, projection='3d')
surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax.contour(X, Y, Z, zdir='z', offset=0, levels=[1], colors='red', linewidths=2, linestyles='--')

# Плоскость z=1
xx, yy = np.meshgrid([x_star - delta, x_star + delta], [y_star - delta, y_star + delta])
zz = np.ones_like(xx)
ax.plot_surface(xx, yy, zz, alpha=0.3, color='red', label='Норма = 1')

# Точка решения
ax.scatter([x_star], [y_star], [norm_phi_prime(x_star, y_star)[0]],
           color='red', s=100, marker='*', label=f'Решение: норм = {norm_phi_prime(x_star, y_star)[0]:.4f}')

ax.set_xlabel('$x$', fontsize=10)
ax.set_ylabel('$y$', fontsize=10)
ax.set_zlabel(r'$||\Phi\'(x,y)||_\infty$', fontsize=10)
ax.set_title('3D визуализация нормы Якобиана', fontsize=12)
ax.legend()

plt.tight_layout()
plt.show()

# Вывод значения нормы в точке решения
norm_at_solution, J_at_solution = norm_phi_prime(x_star, y_star)
print(f"Норма Якобиана в точке решения (x*, y*) = ({x_star:.5f}, {y_star:.5f}):")
print(f"||Φ'(x*, y*)||∞ = {norm_at_solution:.6f}")
print(f"\nМатрица Якобиана в точке решения:")
print(J_at_solution)
print(f"\nПроверка условия сходимости: ||Φ'(x*, y*)||∞ = {norm_at_solution:.6f} < 1")
print("Условие сходимости МПИ выполнено!")

# Анализ влияния радиуса окрестности
print("\n" + "="*60)
print("Анализ нормы в зависимости от расстояния от решения:")
print("="*60)

radii = [0.01, 0.05, 0.1, 0.2]
for r in radii:
    # Проверяем 4 точки на границе окружности радиуса r
    points = [(x_star + r, y_star), (x_star - r, y_star),
              (x_star, y_star + r), (x_star, y_star - r)]
    max_norm = 0
    for point in points:
        norm_val, _ = norm_phi_prime(point[0], point[1])
        max_norm = max(max_norm, norm_val)
    print(f"Радиус {r:.2f}: максимальная норма на границе = {max_norm:.6f}")
```
